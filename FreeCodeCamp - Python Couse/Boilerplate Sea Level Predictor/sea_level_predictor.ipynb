```python
import pandas as pd
import matplotlib.pyplot as plt
from scipy.stats import linregress
```

```python
df = pd.read_csv("epa-sea-level.csv")
```

```python
#Valores
x = df["Year"]
y = df["CSIRO Adjusted Sea Level"]
plt.scatter(x, y)

#Linha
fit1 = linregress(x, y)
y_pred = fit1.intercept + fit1.slope * x #(ax+b)
plt.plot(x, y_pred, color='red', label=f'Best fit line: y={fit1.slope:.2f}x+{fit1.intercept:.2f}')

# Regressão apenas para dados desde 2000
x_recent = df[df["Year"] >= 2000]["Year"]
y_recent = df[df["Year"] >= 2000]["CSIRO Adjusted Sea Level"]
fit2 = linregress(x_recent, y_recent)

# Gera anos de 2000 até 2050
years_extended = pd.Series(range(2000, 2051))
y_pred_extended = fit2.intercept + fit2.slope * years_extended

# Plota a linha de previsão até 2050
plt.plot(years_extended, y_pred_extended, color='green', label='Prediction to 2050')
plt.legend()

# Adicionando rótulos e título
plt.xlabel("Year")
plt.ylabel("Sea Level (inches)")
plt.title("Rise in Sea Level")

plt.show()
```
